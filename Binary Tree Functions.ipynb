```python
tree = [[[7], 1, [9]], 3, [[8], 5, [4]]]
```

```python
def numOfNodes(t):  #returns the number of nodes in a tree
    if len(t) == 1:
        return 1
    else:
        leftNum = numOfNodes(t[0])
        rightNum = numOfNodes(t[2])
        return leftNum + rightNum + 1

print(numOfNodes(tree))
```

```python
def sumOfNodes(t):       #returns the sum of all the nodes in a tree
    if len(t) == 1:
        return t[0]
    else:
        leftSum = sumOfNodes(t[0])
        rightSum = sumOfNodes(t[2])
        return leftSum + rightSum + t[1]
print(sumOfNodes(tree))
```

```python
def maxNode(t):        #returns the max value of a tree
    if len(t) == 1:
        return t[0]
    else:
        maxNum = t[1]
        maxLeft = maxNode(t[0])
        maxRight = maxNode(t[2])
        if maxLeft > maxNum:
            maxNum = maxLeft
        if maxRight > maxNum:
            maxNum = maxRight

        return maxNum

print(maxNode(tree))
```

```python
def minNode(t):         #returns the min value of a tree
    if len(t) == 1:
        return t[0]
    else:
        minNum = t[1]
        minLeft = minNode(t[0])
        minRight = minNode(t[2])
        if minLeft < minNum:
            minNum = minLeft
        if minRight < minNum:
            minNum = minRight

        return minNum

print(minNode(tree))
```

```python
def mirror(t):        #mirrors the tree
    if len(t) == 1:
        return t
    else:
        mirrorLeft = mirror(t[0])
        mirrorRight = mirror(t[2])
        return [mirrorRight, t[1], mirrorLeft]

print(tree)
print(mirror(tree))
```

```python
def drawTree(t, level):
    if len(t) == 1:
        print("   "*level, end = "")
        print(t[0])
    else:
        drawTree(t[2], level + 1)
        print("   "*level, end = "")
        print(t[1])
        drawTree(t[0], level + 1)

drawTree(tree, 0)
print("---------------------------")
drawTree(mirror(tree), 1)
```
