```python
with open('marks.txt', 'r') as marks:
    marks_list = []
    while True:
        line = marks.readline()
        marks_list.append(line)
        if not line:
            break

print(marks_list)
```

```python
with open('marks.txt', 'r') as marks:
    marks_list = []
    for line in marks:
        words = line.split()
        marks_list.append(words)

print(marks_list)
```

```python
import turtle

def setup_screen():
    screen = turtle.Screen()
    screen.bgcolor("lightgrey")
    tess = turtle.Turtle()
    tess.color("blue", "lightblue")
    tess.pensize(3)
    return tess, screen

def draw_axes(t):
    t.penup()
    t.goto(-300, -300)
    t.pendown()
    t.begin_fill()
    t.forward(500)        #x-axis

    t.penup()
    t.goto(-300, -300)
    t.left(90)
    t.pendown()
    t.forward(500)       #y-axis
    t.penup()

    t.goto(-305,-325)
    t.right(90)
    for i in range(11):
        t.write(str(i*10), font=("Times New Roman", 20, "bold"))      #x-axis labels
        t.forward(40)
    t.goto(-300,-300)

def draw_bar(t,height):
    t.begin_fill()
    t.pendown()
    t.left(90)
    t.forward(height*5)
    t.right(90)

    t.forward(10)
    t.write(str(height), font=("Times New Roman", 20, "bold"))
    t.forward(30)

    t.right(90)
    t.forward(height*5)
    t.left(90)

tess, screen = setup_screen()

draw_axes(tess)

for value in ca1_mark_band_count:
    draw_bar(tess,value)

screen.exitonclick()
```
